```python
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI

load_dotenv()

llm = ChatGoogleGenerativeAI(model = "gemma-4-31b-it")
```

```python
with open("Blood_test_report.txt","r") as f:
    blood_test = f.read()
print(blood_test[:200])
```

```python
extraction_prompt = f"""
You are a model data extraction assistant.

From the blood report below, extract all test values amd classify each one as HIGH, Low, or NORMAL based on the reference ranges provided in the report

Format your response as:
- Test Name: value | Status: HIGH/LOW/NORMAL | Reference: range

Blood report:
{blood_test}
"""

extraction_response = llm.invoke(extraction_prompt)
print("=== STAGE 1: Extraction Values ===")
print(extraction_response.text)
```

```python
diet_prompt = f"""
You are a clinical nutritionist specializing in Vegetarian Indian dietry habits.
Based on the blood work analysis below, write:
1. A short health summary in 4-5 lines explaining the patient's condition in simple language
2. A short, practical Indian diet plan having only two section (1) Foods to avoid (2) Foods to eat more of
Blood work analysis :
{extraction_response}
"""
diet_response = llm.invoke(diet_prompt)
print("=== Stge 2: Health Summary and Diet Plan ===")
print(diet_response.text)
```
